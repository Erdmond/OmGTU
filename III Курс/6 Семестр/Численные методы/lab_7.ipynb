```python
import numpy as np
import sympy as sp
```

```python
n = 8
a, b = -2, 2
def f(x): return x**2 + np.sqrt(x + 2) - 1
```

```python
def calc_rect(f, a, b, n):
    h = (b - a) / n
    return h * np.sum(f(np.linspace(a, b, n, endpoint=False)))

res_rect = calc_rect(f, a, b, n)
res_rect
```

```python
def calc_trap(f, a, b, n):
    h = (b - a) / n
    y = f(np.linspace(a, b, n + 1))
    return h * (0.5 * y[0] + np.sum(y[1:-1]) + 0.5 * y[-1])

res_trap = calc_trap(f, a, b, n)
res_trap
```

```python
def calc_simp(f, a, b, n):
    n = n if n % 2 == 0 else n + 1
    h = (b - a) / n
    y = f(np.linspace(a, b, n + 1))
    return h / 3 * (y[0] + 4 * np.sum(y[1:-1:2]) + 2 * np.sum(y[2:-1:2]) + y[-1])

res_sim = calc_simp(f, a, b, n)
res_sim
```

```python
x = sp.symbols('x')
f_sym = x**2 + sp.sqrt(x + 2) - 1
F = sp.integrate(f_sym, (x))
res_exact = float(sp.N(F.subs(x, 2) - F.subs(x, -2)))
res_exact
```

```python
print(f"Ньютон-Лейбниц: {float(res_exact):.8f}")
print(f"Метод прямоугольников: {res_rect:.8f}, ошибка: {abs(res_rect - res_exact):.4}")
print(f"Метод трапеций: {res_trap:.8f}, ошибка: {abs(res_trap - res_exact):.4}")
print(f"Метод Симпсона: {res_sim:.8f}, ошибка: {abs(res_sim - res_exact):.4}")
```
