```python
import numpy as np
import sympy as sp
from IPython.display import display, Latex
from scipy.linalg import solve

sp.init_printing(use_latex='mathjax')
```

```python
f = lambda x, y: 7 - x*y
a, b, h, y0 = -2, 0, 0.5, 0
x = np.linspace(a, b, int((b-a)/h) + 1)

def euler(x, y0, h, f):
    y = np.empty_like(x); y[0] = y0
    for i in range(len(x)-1): y[i+1] = y[i] + h*f(x[i], y[i])
    return y

def mod_euler(x, y0, h, f):
    y = np.empty_like(x); y[0] = y0
    for i in range(len(x)-1):
        yp = y[i] + h*f(x[i], y[i])
        y[i+1] = y[i] + h/2 * (f(x[i], y[i]) + f(x[i+1], yp))
    return y

def rk4(x, y0, h, f):
    y = np.empty_like(x); y[0] = y0
    for i in range(len(x)-1):
        k1 = h*f(x[i], y[i]); k2 = h*f(x[i]+h/2, y[i]+k1/2)
        k3 = h*f(x[i]+h/2, y[i]+k2/2); k4 = h*f(x[i]+h, y[i]+k3)
        y[i+1] = y[i] + (k1 + 2*k2 + 2*k3 + k4)/6
    return y

y_e, y_me, y_rk = euler(x, y0, h, f), mod_euler(x, y0, h, f), rk4(x, y0, h, f)

display(Latex(r"$\mathbf{y'=7-xy,\; y(-2)=0,\; x\in[-2;0],\; h=0.5}$"))
M_cauchy = sp.Matrix(np.round(np.column_stack((x, y_e, y_me, y_rk)), 4))
display(M_cauchy)
```

```python
h2 = 0.1
x2 = np.linspace(0.3, 0.6, int(0.3/h2) + 1)
n = len(x2)
A = np.zeros((n, n)); B = np.zeros(n); rhs = 0.4 * h2**2
al, be = 1, 2

for i in range(n):
    xi = x2[i]
    A[i, i] = -2 - h2**2
    if i == 0:
        A[i, 1] = 2; B[i] = rhs + 2*h2*al*(1 - h2*xi)
    elif i == n-1:
        A[i, n-2] = 2; B[i] = rhs - 2*h2*be*(1 + h2*xi)
    else:
        A[i, i-1] = 1 - h2*xi; A[i, i+1] = 1 + h2*xi; B[i] = rhs

y_bvp = solve(A, B)

display(Latex(r"$\mathbf{y''+2xy'-y=0.4,\; y'(0.3)=1,\; y'(0.6)=2,\; h=0.1}$"))
M_bvp = sp.Matrix(np.round(np.column_stack((x2, y_bvp)), 4))
display(M_bvp)
```
