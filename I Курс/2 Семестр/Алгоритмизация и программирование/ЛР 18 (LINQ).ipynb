Необходимо с использованием запросов отобрать клиентов с отрицательным балансом, посчитать количество клиентов с положительным балансом, выдать клиентов с максимальным доходом, посчитать общую сумму налогов

```polyglot-notebook
public class Account
{
    public int Num { get; set; }
    public string Name { get; set; }
    public int Income { get; set; }
    public int Expence { get; set; }
    public double Tax { get; set; }

    public Account(int nu, string na, int i, int e)
    {
        Num = nu;
        Name = na;
        Income = i;
        Expence = e;
        Tax = i * 0.05;
    }
}
```

```polyglot-notebook
List<Account> accounts = new List<Account>();
```

```polyglot-notebook
var negativeBalanceAccounts =
    from account in accounts
    where account.Income - account.Expence - account.Tax < 0
    select account;
```

```polyglot-notebook
var positiveBalanceCount =
    (from account in accounts
    where account.Income - account.Expence - account.Tax > 0
    select account).Count();
```

```polyglot-notebook
var maxIncomeAccounts = 
    (from account in accounts
     where account.Income == accounts.Max(a => a.Income)
     select account).ToList();
```

```polyglot-notebook
var totalTax =
    (from account in accounts
    select account.Tax).Sum();
```
