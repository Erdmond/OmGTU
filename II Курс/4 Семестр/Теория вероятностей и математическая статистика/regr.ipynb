```python
x = np.array([5, 10, 15, 20, 25, 30, 35, 40, 45, 50])
y_mean = np.array([2.0, 1.9, 2.5, 3.2, 3.6, 4.0, 4.4, 4.4, 3.7, 7.8])

plt.plot(x, y_mean, marker='o', markersize=5, linewidth=2)

plt.xlabel('Сумма счёта (X)')
plt.ylabel('Средние чаевые (Y)')
plt.title('Эмпирическая линия регрессии Y на X')
plt.grid(True)
plt.xticks(x)

for xi, yi in zip(x, y_mean):
    plt.text(xi, yi, f'{yi:.1f}', ha='center', fontweight='bold', bbox=dict(facecolor='white', alpha=0.7, edgecolor='none', boxstyle='round,pad=0.2'))
    
plt.show()
```

```python
from scipy.stats import f

alpha = 0.05       # Уровень значимости
df1 = 1            # Степени свободы числителя (межгрупповые)
df2 = 242          # Степени свободы знаменателя (внутригрупповые)

f_crit = f.ppf(1 - alpha, df1, df2)
print(f"F-критическое (α=0.05, df1=1, df2=242) = {f_crit:.4f}")
```
