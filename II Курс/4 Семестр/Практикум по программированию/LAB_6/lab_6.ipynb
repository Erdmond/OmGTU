## Задание 1. Собственные функции

```python
# Функция статистики числового списка
def analyze_numlers(lst: list[int | float]) -> tuple[int | float]:
    """
    :param lst: Список чисел
    :return: Кортеж числовых характеристик (минимум, максимум, сумма, среднее)
    """
    if not lst:
        return (None, None, None, None)
    
    try:
        return (min(lst), max(lst), sum(lst), sum(lst)/len(lst))
    except Exception as e:
        print(f"Error! {e}")


# Функция частотного анализа символов
def char_frequency(text: str) -> dict[str: int]:
    """
    :param text: Текст
    :return: Словарь уникальных символов без учёта регистра, пробелов и пунктуации
    """
    try:    
        text = text.lower().replace(' ', '')
        text = text.translate(str.maketrans('', '', punctuation))

        return {char: text.count(char) for char in text}
    except Exception as e:
        print(f"Error! {e}")


# Функция фильтрации строк по длине с вргументом по умолчанию
def filter_long_words(words: list[str], min_length: int=5) -> list[str]:
    """
    :param words: Список слов
    :param min_length: Минимальная длина слова для фильтрации
    :return: Список слов, прошедших фильтр
    """
    try:
        return [word for word in words if len(word) > min_length]
    except Exception as e:
        print(f"Error! {e}")
```

## Задание 2. Рекурсия

```python
# Рекурсивное вычисление факториала
def factorial(n: int) -> int:
    """
    :param n: Число
    :return: Факториал числа
    """
    try:
        if n < 0:
            raise ValueError("Impossible to calculate the factorial for a negative number")
        
        match n:
            case 0:
                return 1
            case 1:
                return 1
            case _:
                return factorial(n - 1) * n
    except Exception as e:
        print(f"Error! {e}")


# Рекурсивное вычисление n-го числа Фибоначчи
def fibonacci(n: int) -> int:
    """
    :param n: Число
    :return: n-нное число Фибоначчи
    """
    try:
        if n < 0:
            raise ValueError("Impossible to calculate the fibonacci number with negative index")
        
        match n:
            case 0:
                return 0
            case 1:
                return 1
            case _:
                return fibonacci(n - 2) + fibonacci(n - 1)
    except Exception as e:
        print(f"Error! {e}")


# Рекурсивный разворот строки
def reverse_str(s: str) -> str:
    """
    :param s: Строка
    :return: Развёрнутая строка
    """   
    try:
        if len(s) <= 1:
            return s
        
        s_1, *s_other = s
        
        return reverse_str(''.join(s_other)) + s_1
    except Exception as e:
        print(f"Error! {e}")
```

## Задание 3. Функции высших порядков

```python
# Двойное выполнение функции
def apply_twice(f: callable, x: any):
    """
    :param f: Применяемая функция
    :param x: Элемент, для которого применяется функция
    :return: Элемент, для которого дважды применена функция
    """
    try:
        return f(f(x))
    except Exception as e:
        print(f"Error! {e}")


# Функция преобразования
def custom_map(func: callable, iterable: iter) -> list:
    """
    :param func: Функция
    :param iterable: Итерируемый объект
    :return: Итерируемый объект, в котором каждому элементу применена функция
    """
    try:
        return [func(i) for i in iterable]
    except Exception as e:
        print(f"Error! {e}")


# Фильтр с возможностью инверсии
def custom_filter(predicate: callable, iterable: iter, invert: bool=False) -> list:
    """
    :param predicate: Функция-предикат
    :param iterable: Итерируемый объект
    :param invert: Параметр инверсии
    :return: Список элементов, проходящий условие предиката
    """
    try:
        if invert:
            return [i for i in iterable if not predicate(i)]
        else:
            return [i for i in iterable if predicate(i)]
    except Exception as e:
        print(f"Error! {e}")
```

```python
# Тесты
print(f"apply_twice(lambda x: x + 3, 4) = {apply_twice(lambda x: x + 3, 4)}")
print(f"custom_map(lambda x: x ** 2, [1, 10, 12, 3]) = {custom_map(lambda x: x ** 2, [1, 10, 12, 3])}")
print(f"custom_filter(lambda x: len(x) > 3, ['mom', 'daddy', 'grandma']) = {custom_filter(lambda x: len(x) > 3, ['mom', 'daddy', 'grandma'])}")
```

## Задание 4. Декораторы

```python
# Декоратор подсчёта времени выполнения функции
def time_it(func: callable):
    """
    :param func: Функция, для которой считается время работы
    """
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        stop = time.time()

        print(f"Function {func.__name__} ran for {stop - start:.4f} sec")
        return result
    return wrapper


# Декоратор меморизации
def memorize(func: callable):
    """
    :param func: Функция, для которой кэшируются вызовы с одинаковыми агрументами
    """
    cached_data = {}

    def wrapper(*args):
        if args not in cached_data:
            cached_data[args] = func(*args)
        return cached_data[args]
    return wrapper
```

```python
# Тестовые функции
@time_it
def num_sum(nums: list[int | float]) -> float | int:
    try:
        return sum(nums)
    except Exception as e:
        print(f"Error! {e}")


@memorize
def cached_fibonacci(n: int) -> int:
    return fibonacci(n)


@time_it
def time_fib(n: int) -> int:
    return fibonacci(n)


@time_it
def time_cached_fib(n: int) -> int:
    return cached_fibonacci(n)


# Тесты
print(f"num_sum([i for i in range(10000000)]) = {num_sum([i for i in range(10000000)])}\n")
print(f"time_fib(33) = {time_fib(33)}\n")
print(f"time_cached_fib(33) = {time_cached_fib(33)}")
```

## Задание 5. Основы функционального программирования

```python
# Вспомогательная функция
@memorize
def is_prime(n: int | float) -> bool:
    """
    :param n: Число
    :return: Булево значение проверки числа на принадлежность к множеству простых
    """
    if isinstance(n, float):
        return False

    if n <= 1:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False

    for d in range(3, isqrt(n), 2):
        if n % d == 0:
            return False
    return True


# Обработка списка чисел с функциональными конструкциями
def process_numbers(numbers: list[int | float]) -> int:
    """
    :param numbers: Список чисел
    :return: Произведение возведённых в квадрат простых чисел из списка
    """
    try:
        return reduce(lambda x, y: x * y, map(lambda x: x** 2, filter(is_prime, numbers)))
    except Exception as e:
        print(f"Error! {e}")


# Поиск палиндромов
def find_palindromes(words: list[str]) -> list[str]:
    """
    :param words: Список слов
    :return: Слова из списка, являющиеся палиндромами
    """
    try:
        return list(filter(lambda word: word.lower() == word.lower()[::-1], words))
    except Exception as e:
        print(f"Error! {e}")
```
