```
using System;
using System.IO;
using System.Reflection;

static void GetAssemblyDetails(Assembly assembly)
{
    string result = "";
    Type[] types = assembly.GetTypes();
    foreach (Type type in types)
    {
        result += $"Тип: {type.FullName}\n";
        if (type.IsClass || type.IsInterface)
        {
            MethodInfo[] methods = type.GetMethods(BindingFlags.Public | BindingFlags.Instance | BindingFlags.Static);
            result += "Публичные методы:\n";
            foreach (MethodInfo method in methods)
            {
                result += $" - {method.Name}\n";
                result += $"   Возвращаемое значение: {method.ReturnType.FullName}\n";
                ParameterInfo[] parameters = method.GetParameters();
                result += "   Параметры:\n";
                foreach (ParameterInfo parameter in parameters)
                {
                    result += $"    - Имя: {parameter.Name}\n";
                    result += $"      Тип: {parameter.ParameterType.FullName}\n";
                }
            }
        }
    }
    using (StreamWriter sw = new StreamWriter("result.txt"))
    {
        sw.Write(result);
    }
}
GetAssemblyDetails(Assembly.LoadFrom("core.dll"));
```

```
using System;
using System.IO;
using System.Linq;
using System.Reflection;

static void GetAssemblyDetails(Assembly assembly)
{
    var result = assembly.GetTypes()
        .Where(type => type.IsClass || type.IsInterface)
        .Select(type =>
        {
            var methods = type.GetMethods(BindingFlags.Public | BindingFlags.Instance | BindingFlags.Static);
            return $"Тип: {type.FullName}\n" +
                   "Публичные методы:\n" +
                   string.Join("", methods.Select(m =>
                       $" - {m.Name}\n" +
                       $"   Возвращаемое значение: {m.ReturnType.FullName}\n" +
                       "   Параметры:\n" +
                       string.Join("", m.GetParameters().Select(p =>
                           $"    - Имя: {p.Name}\n" +
                           $"      Тип: {p.ParameterType.FullName}\n"))));
        });
    File.WriteAllLines("result.txt", result);
}

GetAssemblyDetails(Assembly.LoadFrom("core.dll"));
```
