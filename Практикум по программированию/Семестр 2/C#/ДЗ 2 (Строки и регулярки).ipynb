Часть 1.

Провести 2 эксперимента (для string и StringBuilder): 
1. Создать строку длиной 10 символов;
2. Запустить цикл от 0 до 1000000, на каждой итерации i-й(mod10) символ меняется на случайный;
3. Вывести время выполнения.

```polyglot-notebook
using System.Diagnostics;

// string
string s_1 = "0000000000";
Stopwatch sw = Stopwatch.StartNew();
for (int i = 0; i <= 1000000; i++) { s_1 = s_1.Remove(i % 10, 1).Insert(i % 10, "0"); }
sw.Stop();
Console.WriteLine($"Время выполнения для string: {sw.Elapsed}");

// stringbuilder
StringBuilder s_2 = new StringBuilder("0000000000");
sw.Restart();
for (int i = 0; i <= 1000000; i++) { s_2[i % 10] = '1'; }
sw.Stop();
Console.WriteLine($"Время выполнения для StringBuilder: {sw.Elapsed}");
```

Часть 2.

Проверить, что строка содержит время в 24-часовом формате с помощью регулярных выражений.

```polyglot-notebook
string input = "Время 14:25";
if (Regex.IsMatch(input, @"([01]?[0-9]|2[0-3]):[0-5][0-9]")) { Console.WriteLine("Строка содержит время в 24-часовом формате."); }
else { Console.WriteLine("Строка не содержит время в 24-часовом формате."); }
```

Часть 3.

Проверить, что строка содержит корректный адрес электронной почты с помощью регулярных выражений:
1. В никнейме содержатся буквы, цифры, дефисы, нижние подчеркивания
2. Домен может быть любого уровня > 1.

```polyglot-notebook
string input = "example@mail.com.ru.re";
if (Regex.IsMatch(input, @"^[a-zA-Z0-9_.-]+@[a-zA-Z0-9]+\.[a-z]+(\.[a-z]+)+$")) { Console.WriteLine("Строка содержит корректный адрес электронной почты."); }
else { Console.WriteLine("Строка содержит некорректный адрес электронной почты."); }
```
